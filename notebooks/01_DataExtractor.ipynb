Run this notebook to extract the data

Extracts the CSVs we need from data/raw/*.zip into data/processed/extracted/

```python
from pathlib import Path
import zipfile
import pandas as pd

RAW = Path("../data/raw")                    # notebooks/ is one level below the repo root
OUT = Path("../data/processed/extracted")
OUT.mkdir(parents=True, exist_ok =True)
```

```python
long_folder_name = "Survey dataset on mental health in tech professionals from Open Sourcing Mental Health (OSMI) surveys (2017-2021)"
TO_EXTRACT = {
    "2014-raw.zip":      {"survey.csv": "2014.csv"},
    "2016-raw.zip":      {"mental-heath-in-tech-2016_20161114.csv": "2016.csv"},  # 'heath' is the real (typo) name
    "2017-2021-raw.zip": {  f"{long_folder_name}/Survey /2017.csv": "2017.csv",
                            f"{long_folder_name}/Survey /2018.csv": "2018.csv",
                            f"{long_folder_name}/Survey /2019.csv": "2019.csv",
                            f"{long_folder_name}/Survey /2020.csv": "2020.csv",
                            f"{long_folder_name}/Survey /2021.csv": "2021.csv"}
}
```

Extraction loop:

```python
for zip_name, members in TO_EXTRACT.items():
    with zipfile.ZipFile(RAW / zip_name) as zf:
        for member, new_name in members.items():
            target = OUT / new_name
            if target.exists():
                print("Target already exists:", target)
                continue
                
            data = zf.read(member)
            target.write_bytes(data)
            print("Wrote:", target, f"({len(data):,} bytes)")
```

```python
EXPECTED_ROWS = {"2014": 1259, "2016": 1433, "2017": 756,
                 "2018": 417, "2019": 352, "2020": 180, "2021": 131}

for year, n in EXPECTED_ROWS.items():
    df = pd.read_csv(OUT / f"{year}.csv")
    status = "OK" if len(df) == n else "MISMATCH"
    print(year, df.shape, "expected rows:", n, status)
```
