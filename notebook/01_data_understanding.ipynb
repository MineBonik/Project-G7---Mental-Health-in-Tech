```python
import re

def norm(c):
    return re.sub(r"\s+", " ", str(c)).strip().lower()

def find_one(df, text):
    hits = [c for c in df.columns if norm(c).startswith(text)]
    if len(hits) != 1:
        raise ValueError(f"{len(hits)} matches for {text!r}")
    return hits[0]

TEXT = {
    "age": "what is your age",
    "gender": "what is your gender",
    "treatment": "have you ever sought treatment",
    "company_size": "how many employees",
    "tech_employer": "is your employer primarily a tech",
    "benefits": "does your employer provide mental health benefits",
    "anonymity": "is your anonymity protected",
    "employer_discussed": "has your employer ever formally discussed",
    "family_history": "do you have a family history",
    "past_disorder": "have you had a mental health disorder in the past",
    "productivity_affected": "do you believe your productivity",
    "share_friends_family": "how willing would you be to share with friends and family",
}

def build_new(df, year):
    cols, missing = {}, []
    for k, t in TEXT.items():
        try:
            cols[k] = df[find_one(df, t)]
        except ValueError as e:
            missing.append((k, str(e)))
    out = pd.DataFrame(cols)
    out["year"] = year
    return out, missing

MAP14 = {"age": "Age", "gender": "Gender", "country": "Country",
         "treatment": "treatment", "company_size": "no_employees",
         "tech_employer": "tech_company", "benefits": "benefits",
         "anonymity": "anonymity", "family_history": "family_history"}

parts = []
d14 = df14[list(MAP14.values())].rename(columns={v: k for k, v in MAP14.items()})
d14["year"] = 2014
parts.append(d14)

for y, d in [(2016, df16), *yearly.items()]:
    out, missing = build_new(d, y)
    print(y, out.shape, "missing:", missing)
    parts.append(out)

unified = pd.concat(parts, ignore_index=True)
print(unified.shape)
print(unified.groupby("year").size())
print(unified.isna().groupby(unified["year"]).mean().round(2))
```

```python
def search(df, *kws):
    return [c for c in df.columns if all(k in norm(c) for k in kws)]

for name, d in [("2016", df16), *[(str(y), x) for y, x in yearly.items()]]:
    print(name)
    print("  consequences:", [c[:100] for c in search(d, "consequence")])
    print("  interview:", [c[:100] for c in search(d, "interview")])
    print("  country:", [c[:80] for c in search(d, "country")])
    print("  remote:", [c[:80] for c in search(d, "remote")])
    print("  currently:", [c[:100] for c in search(d, "currently")])
print("2014:", [c for c in df14.columns if "consequence" in c or "interview" in c or "remote" in c])
```

```python
import re

def norm(c):
    c = re.sub(r"<[^>]+>", "", str(c))          # remove <strong> tags
    c = c.replace("*", "").replace("\xa0", " ")  # remove markdown stars
    return re.sub(r"\s+", " ", c).strip().lower()

def find_one(df, text):
    hits = [c for c in df.columns if norm(c).startswith(text)]
    if len(hits) != 1:
        raise ValueError(f"{len(hits)} matches for {text!r}")
    return hits[0]

TEXT = {
    "age": "what is your age",
    "gender": "what is your gender",
    "country": "what country do you live in",
    "self_employed": "are you self-employed",
    "treatment": "have you ever sought treatment",
    "company_size": "how many employees",
    "tech_employer": "is your employer primarily a tech",
    "benefits": "does your employer provide mental health benefits",
    "anonymity": "is your anonymity protected",
    "employer_discussed": "has your employer ever formally discussed",
    "family_history": "do you have a family history",
    "past_disorder": "have you had a mental health disorder in the past",
    "current_disorder": "do you currently have a mental health disorder",
    "interview_mental": "would you bring up",
}

def build_new(df, year):
    cols, missing = {}, []
    for k, t in TEXT.items():
        try:
            cols[k] = df[find_one(df, t)]
        except ValueError as e:
            missing.append((k, str(e)))
    out = pd.DataFrame(cols)
    out["year"] = year
    return out, missing

MAP14 = {"age": "Age", "gender": "Gender", "country": "Country",
         "self_employed": "self_employed", "treatment": "treatment",
         "company_size": "no_employees", "tech_employer": "tech_company",
         "benefits": "benefits", "anonymity": "anonymity",
         "family_history": "family_history",
         "interview_mental": "mental_health_interview"}

d14 = df14[list(MAP14.values())].rename(columns={v: k for k, v in MAP14.items()})
d14["year"] = 2014
parts = [d14]
for y, d in [(2016, df16), *yearly.items()]:
    out, missing = build_new(d, y)
    print(y, out.shape, "missing:", missing)
    parts.append(out)

unified = pd.concat(parts, ignore_index=True)
print(unified.shape)
print(unified.isna().groupby(unified["year"]).mean().round(2).T)
```

```python
def to_binary(s):
    m = {"yes": 1, "true": 1, "1": 1, "1.0": 1,
         "no": 0, "false": 0, "0": 0, "0.0": 0}
    return s.astype(str).str.strip().str.lower().map(m)

unified["treatment"] = to_binary(unified["treatment"])
unified["self_employed"] = to_binary(unified["self_employed"])

print(unified.groupby("year")["treatment"].agg(["mean", "count"]).round(3))
print("unmapped:", unified["treatment"].isna().sum(), unified["self_employed"].isna().sum())

# sanity check that 1 really means "yes" in 2016-2021
chk = unified[unified["year"] >= 2016]
print(pd.crosstab(chk["treatment"], chk["current_disorder"].astype(str), normalize="index").round(2))
```

```python
def clean_gender(x):
    if pd.isna(x): return "unknown"
    s = re.sub(r"\s+", " ", str(x)).strip().lower()
    if re.search(r"trans|non-?binary|queer|fluid|agender|enby|neutral|androgyn", s): return "other"
    if re.search(r"\b(female|woman|f|femme)\b", s): return "woman"
    if re.search(r"\b(male|man|m)\b", s): return "man"
    return "other"

gender_raw = unified["gender"].copy()       # kept outside the table
unified["gender"] = gender_raw.map(clean_gender)
print(pd.crosstab(unified["year"], unified["gender"]))
print(gender_raw[unified["gender"] == "other"].str.lower().str.strip().value_counts().head(30))
```

```python
MAN = {"m","male","man","make","mail","maile","malr","mal","dude","masculine",
       "cis male","cis man","cis-male"}
WOMAN = {"f","female","woman","femail","femake","femme","cis female","cis woman","cis-female"}

def clean_gender(x):
    if pd.isna(x): return "unknown"
    s = re.sub(r"\s+", " ", str(x)).strip().lower()
    if s in MAN: return "man"
    if s in WOMAN: return "woman"
    if re.search(r"trans|non-?\s?binary|queer|fluid|agender|enby|androgyn|neutral", s):
        return "other"
    if s.startswith(("fem", "wom")): return "woman"
    if s.startswith(("mal", "guy", "man")): return "man"
    if s in {"nah", "all", "human"}: return "unknown"
    return "other"

unified["gender"] = gender_raw.map(clean_gender)
print(pd.crosstab(unified["year"], unified["gender"]))
print(gender_raw[unified["gender"] == "other"].str.lower().str.strip().value_counts().head(25))
```

```python
ANS = {"yes": "Yes", "no": "No",
       "don't know": "Don't know", "i don't know": "Don't know",
       "not eligible for coverage / n/a": "Not eligible",
       "not eligible for coverage / na": "Not eligible"}

for col in ["benefits", "anonymity"]:
    before = unified[col].isna().sum()
    s = unified[col].astype("string").str.strip().str.lower().str.replace("’", "'", regex=False)
    unified[col] = s.map(ANS)
    print(col, "NaN before:", before, "after:", unified[col].isna().sum())  # must be equal

order = ["1-5", "6-25", "26-100", "100-500", "500-1000", "More than 1000"]
unified["company_size"] = pd.Categorical(unified["company_size"], categories=order, ordered=True)
```

```python
def get_sup(df):
    hits = [c for c in df.columns if "supervisor" in norm(c) and "comfortable" in norm(c)]
    assert len(hits) == 1, hits
    return df[hits[0]]

sup_parts = [df14["supervisor"]] + [get_sup(d) for d in [df16] + list(yearly.values())]
unified["supervisor"] = pd.concat(sup_parts, ignore_index=True).values
print(pd.crosstab(unified["year"], unified["supervisor"].fillna("MISSING")))
```
